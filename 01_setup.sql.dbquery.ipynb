```sql
-- Criando o banco de dados
CREATE DATABASE IF NOT EXISTS db_analytics;

-- tabela bronze, dados brutos do bitcoin
CREATE TABLE IF NOT EXISTS db_analytics.precos_brutos_bitcoin(
    data        DATE,
    abertura    DOUBLE,
    maximo      DOUBLE,
    minimo      DOUBLE,
    fechamento  DOUBLE,
    volume      DOUBLE
)
USING DELTA;

-- tabela gold, dados tratados
CREATE TABLE IF NOT EXISTS db_analytics.precos_semanal_bitcoin(
    semana_inicio DATE,
    semana_fim DATE,
    abertura DOUBLE,
    fechamento DOUBLE,
    maximo DOUBLE,
    minimo DOUBLE,
    variacao_pct DOUBLE,
    volume_medio DOUBLE
)
USING DELTA;

```
